# Shale VLB Simulation Analysis

```
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
import os

# Visualize simulation results
fig, axes = plt.subplots(1, 2, figsize=(18, 7))

if os.path.exists('plots/shale_congestion_report.png'):
    img1 = mpimg.imread('plots/shale_congestion_report.png')
    axes[0].imshow(img1)
    axes[0].axis('off')
    axes[0].set_title('Latency over Time')

if os.path.exists('plots/shale_latency_dist.png'):
    img2 = mpimg.imread('plots/shale_latency_dist.png')
    axes[1].imshow(img2)
    axes[1].axis('off')
    axes[1].set_title('Hop Distribution (h=1 vs h=4)')

plt.tight_layout()
plt.show()
```
